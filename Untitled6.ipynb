```python
class ATM:
    def __init__(self):
        self.balance = 100.00
        self.commands = {
            "1": self.check_balance,
            "2": self.withdraw_money,
            "3": self.deposit_money,
            "4": self.exit_application
        }

    def start_application(self):
        while True:
            self.display_commands()
            command = input("Enter the number of the command you want to perform: ")
            if command in self.commands:
                self.commands[command]()
            else:
                print("Invalid command. Please try again.")

    def display_commands(self):
        print("\nATM Application")
        print("1. Check Balance")
        print("2. Withdraw Money")
        print("3. Deposit Money")
        print("4. Exit Application")

    def check_balance(self):
        print("\nYour current balance is: ", self.balance)

    def withdraw_money(self):
        amount = float(input("\nEnter the amount you want to withdraw: "))
        if amount > self.balance:
            print("\nInsufficient balance. Please try again.")
        else:
            self.balance -= amount
            print("\nYou have withdrawn: ", amount)
            print("Your remaining balance is: ", self.balance)

    def deposit_money(self):
        amount = float(input("\nEnter the amount you want to deposit: "))
        if amount <= 0:
            print("\nInvalid amount. Please try again.")
        else:
            self.balance += amount
            print("\nYou have deposited: ", amount)
            print("Your new balance is: ", self.balance)

    def exit_application(self):
        print("\nThank you for using the ATM application. Goodbye!")
        exit()

if __name__ == "__main__":
    atm = ATM()
    atm.start_application()
```
